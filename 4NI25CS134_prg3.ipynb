```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\user\Downloads\archive (2)\Iris.csv")
```

```python
print("First 5 rows of the dataset:")
print(df.head())
print("\nLast 5 rows of the dataset:")
print(df.tail())
print("\nShape of the dataset:")
print(df.shape)
```

```python
print("\nColumn names:")
print(df.columns)
```

```python
print("\nDataset information:")
df.info()
```

```python
print(df.isnull().sum())
df=df.dropna()
print(df.isnull().sum())
```

```python
print(df.duplicated().sum()) #check duplicates rows
df=df.drop_duplicates()   #removes
print(df.duplicated().sum()) #checks again
```

```python
print(df.describe()) #dispaly statistical summary
sns.set_style("whitegrid") #set graph style
```

```python
plt.figure(figsize=(8,5)) #create a graph area
sns.histplot(data=df, x="SepalLengthCm",bins=10) 
plt.title("Distribution of Sepal Length ")
plt.xlabel("Sepal length (cm)")
plt.ylabel("Number of Flowers")
plt.show() #display graph
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Species",y="SepalLengthCm")
plt.title("Sepal length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,
                x="SepalLengthCm",
                y="PetalLengthCm",
                hue="Species"
               )
plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
sns.pairplot(df,hue="Species") #create a pair plot
plt.show()
```

```python
numeric_columns=[ #select numerical columns
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"]
correlation=df[numeric_columns].corr()
print(correlation)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation ,annot=True,cmap="coolwarm",fmt=".2f") #create correltaion heatmap
plt.title("Correlation Heatmap")
plt.show()
```
